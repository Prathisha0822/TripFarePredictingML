Using Haversine distance function to calculate the distance : 

```python
import numpy as np

def haversine(pickup_lon, pickup_lat, dropoff_lon, dropoff_lat):
    R = 6371  # Earth radius in KM

    pickup_lon, pickup_lat, dropoff_lon, dropoff_lat = map(
        np.radians,
        [pickup_lon, pickup_lat, dropoff_lon, dropoff_lat]
    )

    dlon = dropoff_lon - pickup_lon
    dlat = dropoff_lat - pickup_lat

    a = np.sin(dlat / 2)**2 + \
        np.cos(pickup_lat) * np.cos(dropoff_lat) * np.sin(dlon / 2)**2

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c
```

```python
df['distance_km'] = haversine(
    df['pickup_longitude'],
    df['pickup_latitude'],
    df['dropoff_longitude'],
    df['dropoff_latitude']
)
```

```python
def iqr_filter(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return df[(df[column] >= lower) & (df[column] <= upper)]
```

```python
df = iqr_filter(df, 'total_amount')
```

### Random Forest - Feature Importance

```python
from sklearn.ensemble import RandomForestRegressor
import pandas as pd

TARGET = 'total_amount'

X = df.drop(columns=[TARGET])
y = df[TARGET]

# Ensure all features are numeric
X = pd.get_dummies(X, drop_first=True)
```

```python
datetime_cols = X.select_dtypes(include=['datetime64[ns]', 'datetime64[ns, America/New_York]']).columns
X = X.drop(columns=datetime_cols)
```

```python
X.dtypes
```

```python
# 1) Fit Random Forest
rf = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf.fit(X, y)
```

```python
# 2) Feature importance table
rf_importance = pd.DataFrame({
    'feature': X.columns,
    'importance': rf.feature_importances_
}).sort_values(by='importance', ascending=False)

print("Top 20 Important Features:")
display(rf_importance.head(20))

# 3) Select features (Threshold-based)
THRESHOLD = 0.01   # 1% importance
selected_features = rf_importance[
    rf_importance['importance'] >= THRESHOLD
]['feature'].tolist()

print("\nSelected Features (importance >= 0.01):")
print(selected_features)

# 4) Create selected feature matrix
X_selected = X[selected_features]

print("\nShape Before Selection:", X.shape)
print("Shape After Selection :", X_selected.shape)
```

```python
final_features = [
    'duration_minutes',
    'distance_km',
    'tip_amount'
]
```

```python
X_final = df[final_features]
y = df['total_amount']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_final,
    y,
    test_size=0.2,
    random_state=42
)
```

### Baseline Model — Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))

print(f"Linear Regression MAE : {mae_lr:.2f}")
print(f"Linear Regression RMSE: {rmse_lr:.2f}")
```

### Regularized Models (Ridge & Lasso)

### Ridge

```python
from sklearn.linear_model import Ridge

ridge = Ridge(alpha=1.0)
ridge.fit(X_train, y_train)

y_pred_ridge = ridge.predict(X_test)
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))
rmse_ridge
```

### Lasso 

```python
from sklearn.linear_model import Lasso

lasso = Lasso(alpha=0.01)
lasso.fit(X_train, y_train)

y_pred_lasso = lasso.predict(X_test)
rmse_lasso = np.sqrt(mean_squared_error(y_test, y_pred_lasso))
rmse_lasso
```

### Tree Models

#### Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)

rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
rmse_rf
```

#### Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor

gb = GradientBoostingRegressor(random_state=42)
gb.fit(X_train, y_train)

y_pred_gb = gb.predict(X_test)
rmse_gb = np.sqrt(mean_squared_error(y_test, y_pred_gb))
rmse_gb
```

### Compare Models

```python
pd.DataFrame({
    'Model': ['Linear', 'Ridge', 'Lasso', 'Random Forest', 'Gradient Boosting'],
    'RMSE': [rmse_lr, rmse_ridge, rmse_lasso, rmse_rf, rmse_gb]
}).sort_values(by='RMSE')
```

```python
final_rf = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

final_rf.fit(X_final, y)
```
